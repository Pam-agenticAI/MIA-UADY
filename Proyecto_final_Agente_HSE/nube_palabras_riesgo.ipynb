# Nube de palabras - Safety Save (MID1)

Genera dos nubes de palabras (riesgo alto y riesgo bajo) a partir del texto de las observaciones,
usando la misma tokenizacion y exclusion de conectores del analisis exploratorio previo.

Requisitos: `pip install wordcloud` (si no esta instalado en tu entorno de Colab) y tener
`safety_save_limpio.csv` en la misma carpeta o montado desde Drive.


```python
!pip install wordcloud --quiet
```

```python
import pandas as pd
import re
from collections import Counter
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Cargar el dataset limpio (debe estar en la misma carpeta de Colab)
df = pd.read_csv('safety_save_limpio.csv')

def tokenize(text):
    return re.findall(r'\b[a-zA-Z]{3,}\b', text.lower())

skip = {'the', 'and', 'for', 'are', 'were'}

alto_words = Counter()
bajo_words = Counter()

for _, row in df.iterrows():
    tokens = [w for w in tokenize(row['texto']) if w not in skip]
    if row['riesgo'] == 'alto':
        alto_words.update(tokens)
    else:
        bajo_words.update(tokens)

# Generar las dos nubes de palabras
wc_alto = WordCloud(
    width=700, height=500, background_color='white',
    colormap='Reds', prefer_horizontal=1.0, max_words=60
).generate_from_frequencies(alto_words)

wc_bajo = WordCloud(
    width=700, height=500, background_color='white',
    colormap='Greens', prefer_horizontal=1.0, max_words=60
).generate_from_frequencies(bajo_words)

# Mostrarlas lado a lado
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

axes[0].imshow(wc_alto, interpolation='bilinear')
axes[0].axis('off')
axes[0].set_title('Riesgo ALTO', fontsize=16)

axes[1].imshow(wc_bajo, interpolation='bilinear')
axes[1].axis('off')
axes[1].set_title('Riesgo BAJO', fontsize=16)

plt.tight_layout()
plt.savefig('nube_palabras_riesgo.png', dpi=150, bbox_inches='tight')
plt.show()
```
