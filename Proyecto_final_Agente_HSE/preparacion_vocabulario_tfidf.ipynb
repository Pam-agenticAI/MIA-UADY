# Preparacion del vocabulario para TF-IDF - Safety Save (MID1)

Calcula las medidas descriptivas basicas del texto (longitud, tamano de vocabulario) y el
conteo de frecuencia completo, sin filtrar por longitud de palabra. Este analisis es el
que justifica la decision de tokenizar solo palabras de 3+ letras y de no usar la lista
estandar de stopwords en ingles (para conservar la negacion "not").

Asegurate de subir `safety_save_limpio.csv` a esta misma carpeta de Colab (o montar
Drive) antes de correr la celda.


```python
import pandas as pd
import re
from collections import Counter

# Cargar el dataset limpio (debe estar en la misma carpeta de Colab)
df = pd.read_csv('safety_save_limpio.csv')

# --- Medidas descriptivas basicas del texto ---
def tokenize_all(text):
    # Sin filtro de longitud, para no descartar nada antes de ver los resultados
    return re.findall(r'\b[a-zA-Z]+\b', text.lower())

lengths = []
all_tokens = []
for texto in df['texto']:
    tokens = tokenize_all(texto)
    all_tokens.extend(tokens)
    lengths.append(len(tokens))

vocab = Counter(all_tokens)

print("Total de observaciones:", len(df))
print("Longitud promedio de cada observacion (palabras):", sum(lengths) / len(lengths))
print("Longitud minima / maxima:", min(lengths), "/", max(lengths))
print("Vocabulario unico total (sin filtrar):", len(vocab))
print()

# --- Palabras que aparecen una sola vez (ruido por rareza) ---
solo_una_vez = sum(1 for w, c in vocab.items() if c == 1)
print(f"Palabras que aparecen en solo 1 observacion: {solo_una_vez} de {len(vocab)}")

dos_o_mas = sum(1 for w, c in vocab.items() if c >= 2)
print(f"Palabras que aparecen en 2+ observaciones: {dos_o_mas}")
print()

# --- Palabras mas frecuentes (ruido por ser conectores) ---
print("Top 20 palabras mas frecuentes:")
for palabra, cuenta in vocab.most_common(20):
    print(f"  {palabra}: {cuenta}")
```
