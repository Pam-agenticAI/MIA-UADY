### Mutabilidad, alias y copia

Dos nombres pueden apuntar a la misma lista. Para modificar el contenedor por separado, debemos copiarlo.
Si contiene otros objetos mutables, una copia superficial aún puede compartirlos. Este detalle explica muchos errores silenciosos
en pipelines de datos.

```python
original = ["dato-a", "dato-b"]
alias = original
copia = original.copy()

alias.append("dato-c")
copia.append("dato-independiente")

print("original:", original)
print("alias:   ", alias)
print("copia:   ", copia)
print("¿Mismo objeto?", original is alias, original is copia)
```

### Copiar una estructura anidada

`.copy()` hace una **copia superficial**: crea el contenedor externo, pero conserva
las referencias a sus elementos. Si dentro hay una lista, esa lista sigue compartida.

```python
registro_original = {"texto": "hola", "etiquetas": ["saludo"]}
registro_copia = registro_original.copy()
registro_copia["etiquetas"].append("corto")
print(registro_original)
print(registro_original is registro_copia)
print(registro_original["etiquetas"] is registro_copia["etiquetas"])
```

### Ejercicio 3 · Copiar sin compartir la lista

Crea una copia que permita añadir `corto` sin cambiar las etiquetas del original. Para esta estructura basta copiar también la lista interna. **Profundización:** investiga qué cambiaría si hubiera más niveles.

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
fuente = {"texto": "hola", "etiquetas": ["saludo"]}
# Construye una copia independiente para esta estructura.

# Primero copio el diccionario por fuera con .copy().
# Pero eso no basta, porque la lista de etiquetas seguiría compartida
# (eso fue lo que pasó en el ejemplo de arriba).
copia = fuente.copy()

# Por eso también copio la lista de adentro y la pongo en la copia.
# Así la copia tiene su propia lista de etiquetas.
copia["etiquetas"] = fuente["etiquetas"].copy()

# Ahora agrego "corto" solo a la copia.
copia["etiquetas"].append("corto")

print("fuente:", fuente)
print("copia: ", copia)

# Compruebo que el original no cambió.
assert fuente["etiquetas"] == ["saludo"]

# Compruebo que la copia sí tiene la etiqueta nueva.
assert copia["etiquetas"] == ["saludo", "corto"]

# Compruebo que ya no comparten la misma lista.
# "is" pregunta si son el mismo objeto, y debe dar False.
assert fuente["etiquetas"] is not copia["etiquetas"]

# Profundización:
# Si la estructura tuviera más niveles (por ejemplo, una lista dentro de otra
# lista dentro del diccionario), tendría que copiar cada nivel a mano,
# y es fácil que se me olvide alguno.
# Para eso existe copy.deepcopy(), que copia todos los niveles de una vez:
import copy

fuente_profunda = {"texto": "hola", "etiquetas": [["saludo", "inicio"]]}
copia_profunda = copy.deepcopy(fuente_profunda)
copia_profunda["etiquetas"][0].append("corto")

# El original no cambia aunque modifiqué un nivel más adentro.
assert fuente_profunda["etiquetas"][0] == ["saludo", "inicio"]
print("deepcopy original:", fuente_profunda)
print("deepcopy copia:   ", copia_profunda)
```

### Diccionarios: representar registros

Un diccionario asocia claves a valores. `in` comprueba las claves; `.items()`
permite recorrer pares clave–valor. `get` usa su valor por defecto cuando la clave
no existe, pero conserva `None` si ese es el valor guardado.

```python
ficha = {"texto": "Excelente", "score": None}
ficha["etiqueta"] = "positivo"
print("score" in ficha)
print(ficha.get("score", 0), ficha.get("idioma", "es"))
for clave, valor in ficha.items():
    print(clave, valor)
```

### Conjuntos: pertenencia, coincidencias y diferencias

Un conjunto conserva valores únicos. Usa `set()` para el conjunto vacío: `{}` crea
un diccionario. La intersección `&` encuentra coincidencias, la unión `|` reúne
valores y la diferencia `-` encuentra los que están sólo en el primer conjunto.
No uses la posición ni el orden impreso de un conjunto como parte del resultado.

```python
permitidas = {"positivo", "neutral", "negativo"}
observadas = {"positivo", "positivo", "desconocido"}
print("Reconocidas:", observadas & permitidas)
print("No reconocidas:", observadas - permitidas)
```

### Ejercicio 4 · Elegir una estructura

Representa dos predicciones con texto, etiqueta y score. Conserva el orden de llegada y encuentra las etiquetas distintas. Explica por qué conviene una lista de diccionarios para los registros y un conjunto para las etiquetas.

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
# Crea dos registros, guárdalos en una lista y reúne sus etiquetas en un set.
# Creo cada predicción como un diccionario, porque así cada dato tiene su nombre
# (texto, etiqueta y score) y no tengo que acordarme de posiciones.
prediccion_1 = {"texto": "Excelente servicio", "etiqueta": "positivo", "score": 0.97}
prediccion_2 = {"texto": "Llegó tarde el pedido", "etiqueta": "negativo", "score": 0.88}

# Las guardo en una lista porque la lista conserva el orden en que llegaron.
predicciones = [prediccion_1, prediccion_2]

# Recorro la lista para ver las predicciones en orden.
for prediccion in predicciones:
    print(prediccion["texto"], "->", prediccion["etiqueta"], prediccion["score"])

# Reúno las etiquetas en un set para quedarme solo con las distintas.
# Si hubiera dos "positivo", el set lo guardaría una sola vez.
etiquetas_distintas = {prediccion["etiqueta"] for prediccion in predicciones}
print("Etiquetas distintas:", etiquetas_distintas)

# Comprobaciones:
# La lista conserva el orden: la primera es la que llegó primero.
assert predicciones[0]["texto"] == "Excelente servicio"
assert predicciones[1]["texto"] == "Llegó tarde el pedido"

# El set tiene las dos etiquetas distintas.
assert etiquetas_distintas == {"positivo", "negativo"}

# Explicación:
# Uso una lista de diccionarios para los registros porque cada predicción
# tiene varios datos con nombre (por eso diccionario) y necesito saber
# en qué orden llegaron (por eso lista). Es como una bitácora de inspecciones:
# cada renglón tiene sus campos y el orden cronológico importa.
# Uso un conjunto para las etiquetas porque solo me interesa saber qué etiquetas
# distintas aparecieron, sin repetidas y sin importar el orden.
# Es como el catálogo de tipos de hallazgo: no importa cuántas veces salió cada uno,
# sino cuáles hubo.
```

## 3. Decisiones y truthiness

En una condición, Python interpreta como falsos `False`, `None`, cero y las
colecciones vacías. Esto permite escribir condiciones directas, siempre que el
significado sea claro.

```python
respuesta_modelo = ""

if not respuesta_modelo:
    estado = "sin respuesta"
elif len(respuesta_modelo) < 20:
    estado = "respuesta breve"
else:
    estado = "respuesta disponible"

print(estado)
```

### Bloques, ciclos y decisiones

Los dos puntos abren un bloque y la sangría indica qué instrucciones pertenecen a
él. Usa cuatro espacios por nivel. `for` repite el bloque para cada elemento;
`range(3)` produce los valores 0, 1 y 2. `continue` salta al siguiente elemento y
`break` termina el ciclo. `while` repite mientras una condición sea verdadera.

```python
for valor in [0.0, None, 0.85]:
    if valor is None:
        continue
    print("Score presente:", valor)

intentos = 0
while intentos < 3:
    intentos += 1
    print("Intento", intentos)
```

### Ejercicio 5 · Cero no es ausencia

Recorre `[None, 0.0, 0.4, 0.95, 0.99]`. Ignora `None` y detente en el primer score mayor o igual a 0.9. Deben mostrarse 0.0, 0.4 y 0.95. ¿Qué se perdería usando `if not score`?

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
valores_busqueda = [None, 0.0, 0.4, 0.95, 0.99]
# Usa for, continue y break.

# Creo una lista vacía para ir guardando los valores que muestro.
# Así después puedo comprobar que salieron los correctos.
mostrados = []

for score in valores_busqueda:
    # Si el valor es None (no hay dato), lo salto con continue.
    # Uso "is None" y no "if not score" para no saltarme el 0.0.
    if score is None:
        continue

    # Si llegó aquí, el valor sí existe, así que lo muestro y lo guardo.
    print("Score:", score)
    mostrados.append(score)

    # Si el score es mayor o igual a 0.9, me detengo con break.
    if score >= 0.9:
        break

# Compruebo que se mostraron exactamente 0.0, 0.4 y 0.95.
# El 0.99 no sale porque el ciclo se detuvo en 0.95.
assert mostrados == [0.0, 0.4, 0.95]

# Ahora pruebo qué pasa si uso "if not score" para ver la diferencia.
mostrados_mal = []
for score in valores_busqueda:
    if not score:
        continue
    mostrados_mal.append(score)
    if score >= 0.9:
        break

print("Con 'if not score':", mostrados_mal)

# Aquí el 0.0 se perdió, porque Python considera el cero como "falso".
assert mostrados_mal == [0.4, 0.95]

# Respuesta:
# Si uso "if not score", Python se salta tanto el None como el 0.0,
# porque los dos cuentan como "falso". Así perdería el 0.0, que sí es
# un dato real: el modelo dio un score de cero.
# Por eso uso "if score is None", que solo salta cuando de verdad no hay dato.
# En mi trabajo sería como descartar una medición de 0 mg/L pensando que
# no se tomó la muestra: estaría eliminando un resultado válido.
```

## 4. Ciclos y comprehensions

Un `for` recorre los elementos de un iterable. En una lista puedes recorrer los
valores directamente; no hace falta construir sus índices. Usa `enumerate` cuando
también necesites la posición y `zip` para avanzar por varias secuencias a la vez.
Por defecto, `zip` se detiene al agotarse la secuencia más corta. Si necesitas
comprobar que tengan la misma longitud, usa `zip(..., strict=True)`.

```python
textos = ["muy bueno", "regular", "excelente"]
scores = [0.82, 0.51, 0.96]

for posicion, (texto, score) in enumerate(zip(textos, scores), start=1):
    print(f"{posicion}. {texto:<12} → {score:.0%}")
```

Las comprehensions sirven para transformaciones pequeñas y legibles. Si la
lógica requiere varias decisiones o efectos secundarios, un ciclo explícito
suele ser mejor.

```python
scores_altos = [score for score in scores if score >= 0.80]
longitudes = {texto: len(texto) for texto in textos}
iniciales = {texto[0] for texto in textos}

print(scores_altos)
print(longitudes)
print(iniciales)
```

### Ejercicio 6 · Transformar y filtrar

Convierte a Fahrenheit las temperaturas mayores o iguales a 25. Usa la fórmula `fahrenheit = celsius * 9 / 5 + 32`. Primero usa un ciclo y después una comprehension. El resultado debe ser `[77.0, 84.2, 87.8]`.

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
temperaturas = [18, 21, 25, 29, 31]
# Construye ambas versiones.

# Versión 1: con un ciclo for.
# Creo una lista vacía para ir guardando las temperaturas convertidas.
fahrenheit_ciclo = []

for celsius in temperaturas:
    # Solo convierto las que son mayores o iguales a 25.
    if celsius >= 25:
        fahrenheit = celsius * 9 / 5 + 32
        fahrenheit_ciclo.append(fahrenheit)

print("Con ciclo:", fahrenheit_ciclo)

# Versión 2: con una comprehension.
# Hace lo mismo que el ciclo de arriba, pero en una sola línea:
# "convierte cada celsius de la lista, solo si es mayor o igual a 25".
fahrenheit_comprehension = [celsius * 9 / 5 + 32 for celsius in temperaturas if celsius >= 25]

print("Con comprehension:", fahrenheit_comprehension)

# Compruebo que las dos versiones dan el resultado que pide el ejercicio.
assert fahrenheit_ciclo == [77.0, 84.2, 87.8]
assert fahrenheit_comprehension == [77.0, 84.2, 87.8]

# Y comprobar que las dos versiones dan exactamente lo mismo.
assert fahrenheit_ciclo == fahrenheit_comprehension

# Nota: 18 y 21 no aparecen porque son menores a 25.
# Los resultados salen con decimales (77.0 y no 77) porque la división "/"
# siempre da un número decimal en Python.
# La comprehension es más corta y se lee bien para algo sencillo como esto.
# Si la lógica fuera más complicada, con varias condiciones,
# preferiría el ciclo porque es más fácil de leer y revisar.
```

## 5. Funciones: nombrar una transformación

Una función bien enfocada reduce repetición y permite probar una idea de forma
aislada. Sus parámetros son entradas; `return` produce una salida. Una función
sin `return` explícito devuelve `None`.

```python
def normalizar_texto(texto, *, minusculas=True):
    '''Elimina espacios externos y opcionalmente convierte a minúsculas.'''
    resultado = texto.strip()
    if minusculas:
        resultado = resultado.lower()
    return resultado


print(normalizar_texto("  Hola, IA  "))
print(normalizar_texto("  Hola, IA  ", minusculas=False))
```

El `*` vuelve explícitos los argumentos posteriores: debemos escribir
`minusculas=False`. En la semana 2 añadiremos anotaciones de tipo y Pydantic;
hoy nos concentramos en el comportamiento.

### Verificar una función con `assert`

`assert condición` comprueba una expectativa. Si es falsa, produce un
`AssertionError`. Lo usaremos para comprobar ejercicios; las entradas externas
se validarán con condiciones y excepciones explícitas.

### Ejercicio 7 · Promedio sin inventar datos

Escribe `promedio_o_none(valores)`: devuelve el promedio o `None` si la lista está vacía. Comprueba `[0.0]`, `[]` y `[0.5, 1.0]`. ¿Por qué cero y ausencia necesitan resultados distintos?

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
# Define promedio_o_none y escribe tres comprobaciones.

def promedio_o_none(valores):
    '''Devuelve el promedio de la lista, o None si la lista está vacía.'''
    # Si la lista no tiene elementos, no hay datos para promediar.
    # En lugar de inventar un número, devuelvo None.
    # Uso len(valores) == 0 para que quede claro que reviso si está vacía.
    if len(valores) == 0:
        return None

    # Si hay datos, calculo el promedio: suma entre cantidad.
    return sum(valores) / len(valores)


# Comprobación 1: una lista con un solo cero.
# Sí hay un dato (el cero), así que el promedio es 0.0.
assert promedio_o_none([0.0]) == 0.0

# Comprobación 2: una lista vacía.
# No hay datos, así que debe devolver None.
# Uso "is None" para comprobarlo, como en el ejercicio 5.
assert promedio_o_none([]) is None

# Comprobación 3: dos valores normales.
# (0.5 + 1.0) / 2 = 0.75
assert promedio_o_none([0.5, 1.0]) == 0.75

print("Promedio de [0.0]:", promedio_o_none([0.0]))
print("Promedio de []:", promedio_o_none([]))
print("Promedio de [0.5, 1.0]:", promedio_o_none([0.5, 1.0]))

# Respuesta:
# Cero y ausencia necesitan resultados distintos porque significan cosas distintas.
# Si la lista tiene [0.0], sí se midió algo y el resultado fue cero.
# Si la lista está vacía, no se midió nada, y no puedo decir que el promedio es cero
# porque estaría inventando un dato.
# Además, si intento dividir entre cero (len de una lista vacía es 0),
# Python marcaría un error (ZeroDivisionError).
# En mi trabajo sería como reportar "promedio mensual de 0 mg/L" en un mes
# en el que no se hizo ningún muestreo: el reporte diría que todo está limpio
# cuando en realidad no hay información.
```

### Ordenar usando una función

`sorted` crea una lista nueva; `.sort()` modifica la lista existente. El argumento
`key` recibe una función que calcula el criterio de orden. Pasamos la función
sin llamarla: Python la aplicará a cada elemento.

```python
evaluaciones = [
    {"modelo": "a", "score": 0.7},
    {"modelo": "b", "score": 0.95},
    {"modelo": "c", "score": 0.8},
]

def obtener_score(evaluacion):
    return evaluacion["score"]

ranking = sorted(evaluaciones, key=obtener_score, reverse=True)
print(ranking)
# lambda permite expresar una función pequeña en el mismo lugar.
assert ranking == sorted(evaluaciones, key=lambda e: e["score"], reverse=True)
```

## 8. POO: objetos que conservan estado

Una **clase** define cómo construir objetos y qué operaciones ofrecen. Cada
**instancia** es un objeto concreto. Sus **atributos** guardan datos y sus
**métodos** son funciones asociadas al objeto.

`__init__` inicializa una instancia; `self` representa esa instancia dentro de los
métodos. Al llamar `contador.agregar()`, Python pasa `contador` como `self`.

```python
class Contador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.total = 0

    def agregar(self):
        self.total += 1

    def reiniciar(self):
        self.total = 0

contador_a = Contador("lote A")
contador_b = Contador("lote B")
contador_a.agregar()
contador_a.agregar()
print(contador_a.nombre, contador_a.total)
print(contador_b.nombre, contador_b.total)
assert contador_b.total == 0
```

### Ejercicio 10 · Estado independiente

Crea `RegistroIndependiente` con una lista por instancia y un método `agregar`. Comprueba que agregar un dato a A deja vacía la lista de B.

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
# Mueve la inicialización de la lista a __init__.
class RegistroIndependiente:
    def __init__(self):
        # Creo la lista dentro de __init__ y no en el cuerpo de la clase.
        # Así cada instancia recibe su propia lista vacía cuando se crea.
        self.registros = []

    def agregar(self, texto):
        # Agrego el texto solo a la lista de esta instancia (self).
        self.registros.append(texto)


# Creo dos registros distintos.
registro_a = RegistroIndependiente()
registro_b = RegistroIndependiente()

# Agrego un dato solo al registro A.
registro_a.agregar("dato de A")

print("registro_a:", registro_a.registros)
print("registro_b:", registro_b.registros)

# Compruebo que A tiene su dato.
assert registro_a.registros == ["dato de A"]

# Compruebo que B sigue vacío.
assert registro_b.registros == []

# Compruebo que no comparten la misma lista.
assert registro_a.registros is not registro_b.registros

# Explicación:
# En RegistroCompartido la lista estaba escrita en el cuerpo de la clase,
# entonces pertenecía a la clase y todas las instancias usaban la misma lista.
# Por eso el dato de A aparecía en B.
# En RegistroIndependiente creé la lista dentro de __init__ con self.registros,
# así que cada instancia tiene su propia lista y los datos no se mezclan.
# Es el mismo problema de los alias del ejercicio 3, pero dentro de una clase:
# el programa no marca error, pero mezcla datos que deberían estar separados.
```

## 9. Caso integrador: un registro de predicciones

La clase siguiente conserva predicciones ya normalizadas y combina estructuras de
datos con comportamiento y estado. Cada registro recibe su posición al agregarse.
`resumen` devuelve cantidad, conteos y promedio; el promedio es `None` cuando no
hay registros.


```python
predicciones = [
    {"texto": "Excelente", "etiqueta": "positivo", "score": 0.98},
    {"texto": "Regular", "etiqueta": "neutral", "score": 0.66},
    {"texto": "Llegó roto", "etiqueta": "negativo", "score": 0.93},
    {"texto": "Cero válido", "etiqueta": "neutral", "score": 0.0},
]
```

```python
class RegistroPredicciones:
    def __init__(self, nombre):
        self.nombre = nombre
        self._registros = []

    def agregar(self, registro, posicion):
        guardado = registro.copy()
        guardado["posicion"] = posicion
        self._registros.append(guardado)

    def resumen(self):
        scores = [registro["score"] for registro in self._registros]
        return {
            "cantidad": len(self._registros),
            "conteos": dict(Counter(r["etiqueta"] for r in self._registros)),
            "promedio": sum(scores) / len(scores) if scores else None,
        }

lote = RegistroPredicciones("Demostración")
for posicion, registro in enumerate(predicciones):
    lote.agregar(registro, posicion)
print(lote.resumen())


```

y promedio **0.6425**. Este promedio describe los scores; no mide exactitud del
modelo porque no tenemos etiquetas de referencia.
